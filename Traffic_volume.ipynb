```python
# import libraries

import numpy as np
import pandas as pd
import glob
import os
```

```python
# maxmimum vol 

# set folder path
folder_path = "C:\\Users\\ashle\\Downloads\\VDOT_traffic_files\\Traffic"

# get a list of all csv files 
files = glob.glob(os.path.join(folder_path, "*Volume*.csv"))

high_vol_list = []
low_vol_list = []
mid_vol_list = []

for file in files:
    traffic_df = pd.read_csv(file)
    high_vol, low_vol, mid_vol = np.percentile(traffic_df['VOL'], 75), np.percentile(traffic_df['VOL'], 25), np.percentile(traffic_df['VOL'], 50)

    # add to respective lists
    high_vol_list.append(high_vol)
    low_vol_list.append(low_vol)
    mid_vol_list.append(mid_vol)

high_traffic_vol = np.mean(high_vol_list)

low_traffic_vol = np.mean(low_vol_list)

mid_traffic_vol = np.mean(mid_vol_list)

print(high_traffic_vol, low_traffic_vol, mid_traffic_vol)
```
